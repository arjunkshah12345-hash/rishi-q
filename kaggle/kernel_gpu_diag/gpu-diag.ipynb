```python
import os, json, platform, subprocess, sys
from pathlib import Path
out = Path('/kaggle/working')
def run(cmd):
    try:
        return subprocess.check_output(cmd, stderr=subprocess.STDOUT, text=True, timeout=60)
    except Exception as e:
        return f'ERR:{e}'
diag = {
  'python': platform.python_version(),
  'platform': platform.platform(),
  'env_nvidia': os.environ.get('NVIDIA_VISIBLE_DEVICES'),
  'env_cuda': os.environ.get('CUDA_VISIBLE_DEVICES'),
  'nvidia_smi': run(['bash','-lc','nvidia-smi 2>&1 || true']),
  'ls_nvidia': run(['bash','-lc','ls -la /dev/nvidia* 2>&1 || true']),
}
try:
  import torch
  diag['torch'] = torch.__version__
  diag['torch_cuda'] = torch.version.cuda
  diag['cuda_available'] = torch.cuda.is_available()
except Exception as e:
  diag['torch_err'] = str(e)
(out/'diag.json').write_text(json.dumps(diag, indent=2))
print(json.dumps(diag, indent=2)[:2000])
print('OK wrote diag.json')
```
